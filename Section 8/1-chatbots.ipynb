```python
from dotenv import load_dotenv
import os
import uvicorn

os.environ["OPEN_API_KEY"] = os.getenv("OPEN_API_KEY", "")
```

```python
from langchain_openai import ChatOpenAI
from pydantic import SecretStr

model = ChatOpenAI(model="gpt-4o-mini")
```

```python
from langchain_core.messages import HumanMessage

model.invoke(("human","Hi I am Saksham Gupta"))
```

```python
from langchain_core.messages import AIMessage

model.invoke(
    [
        ("human", "Hi I am Saksham Gupta I am a Chief AI Engineer"),
        ("ai", "Hello Saksham! How can I assist you today?"),
        ("human", "hey what my name and what do I do?")
    ]
)
```

```python
## Message History

from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store = {}


def get_session_history(session_id: str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]


with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
from langchain_core.runnables import RunnableConfig


config = RunnableConfig({"configurable": {"session_id": "chat1"}})
```

```python
response = with_message_history.invoke(
    [HumanMessage(content="Hi I am Saksham Gupta and I am a AI Engineer")],
    config=config,
)

print(response)
```

```python
response = with_message_history.invoke(
    [HumanMessage(content="What is my name")],
    config=config,
)

print(response)
```

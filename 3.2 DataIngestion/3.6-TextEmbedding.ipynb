```python
import os 
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
embeddings
```

```python
text = "This is a tutorial on OPENAI embedding"
query_result = embeddings.embed_query(text)
query_result
```

```python
from langchain_community.document_loaders import TextLoader

loader = TextLoader("speech.txt")

docs = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=10)

final_document = text_splitter.split_documents(docs)

final_document
```

```python
from langchain_community.vectorstores import Chroma
# Embedding + VectorDB

embeddings1024 = OpenAIEmbeddings(model="text-embedding-3-small", dimensions=1024)
db = Chroma.from_documents(final_document, embeddings1024)
db
```

```python
retrieved_results = db.similarity_search("from a lawless and malignant few.It is a distressing and oppressive")

print(retrieved_results)
```
